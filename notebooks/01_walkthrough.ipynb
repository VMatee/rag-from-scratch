# BM25 retrieval and evidence-grounded answers

This notebook implements the *retrieval* layer of RAG from the Python standard library. Optional Ollama generation is not used here, so the result is **not** a language-model answer.

```python
from pathlib import Path
from rag_lab import BM25Index, load_documents, format_context

path = Path("../data/documents.jsonl")
if not path.exists():
    path = Path("data/documents.jsonl")
index = BM25Index(load_documents(path))
print("Indexed documents:", len(index.documents))
```

## Rank passages for a query

BM25 uses token frequencies, inverse document frequency and document-length normalization. Inspect the numeric scores and retained source identifiers.

```python
query = "How do I record changes in Git?"
hits = index.search(query, top_k=3)
for hit in hits:
    print(hit.doc.id, round(hit.score, 3), hit.doc.source)
print(format_context(hits))
```

## Failure modes

Term-based retrieval is brittle for unseen synonyms, language changes, and missing documents.

```python
print(index.search("quantum bananas", top_k=3))
assert index.search("quantum bananas", top_k=3) == []
```
